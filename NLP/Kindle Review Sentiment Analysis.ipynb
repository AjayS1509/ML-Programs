```python
def change_review(value):
    if(value < 3):
        return 0
    else: return 1
```

```python
##Preprocessing and Cleaning
df['rating'] = df['rating'].apply(lambda x:0 if x < 3 else 1) ##positive review is 1 and negative is 0
```

```python
##lower all the cases
df['reviewText'] = df['reviewText'].str.lower()
```

```python
df.reviewText.head()
```

```python
import re
from nltk.corpus import stopwords
from bs4 import BeautifulSoup
##removing special charaters
df['reviewText']=df['reviewText'].apply(lambda x:re.sub('[^a-z A-z 0-9-]+', '',x))
## Remove the stopswords
df['reviewText']=df['reviewText'].apply(lambda x:" ".join([y for y in x.split() if y not in stopwords.words('english')]))
## Remove url 
df['reviewText']=df['reviewText'].apply(lambda x: re.sub(r'(http|https|ftp|ssh)://([\w_-]+(?:(?:\.[\w_-]+)+))([\w.,@?^=%&:/~+#-]*[\w@?^=%&/~+#-])?', '' , str(x)))
## Remove html tags
df['reviewText']=df['reviewText'].apply(lambda x: BeautifulSoup(x, 'lxml').get_text())
## Remove any additional spaces
df['reviewText']=df['reviewText'].apply(lambda x: " ".join(x.split()))
```

```python
df.head()
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
def lemmatize_word(text):
    return " ".join([lemmatizer.lemmatize(word) for word in text.split()])
```

```python
df['reviewText']=df['reviewText'].apply(lambda x: lemmatize_word(x))
```

```python
df.head()
```

```python
##Train test split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df['reviewText'],df['rating'],test_size=0.2, random_state=42)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
bow = CountVectorizer()
```

```python
X_train_bow = bow.fit_transform(X_train).toarray()
```

```python
X_test_bow = bow.transform(X_test).toarray()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer()
```

```python
X_train_tfidf = tfidf.fit_transform(X_train).toarray()
X_test_tfidf = tfidf.transform(X_test).toarray()
```

```python
X_train_bow
```

```python
from sklearn.naive_bayes import GaussianNB
nb_mobel_bow = GaussianNB().fit(X_train_bow, y_train)
nb_mobel_tfidf = GaussianNB().fit(X_train_tfidf, y_train)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
y_pred_nbow = nb_mobel_bow.predict(X_test_bow)
```

```python
y_pred_tfidf = nb_mobel_tfidf.predict(X_test_tfidf)
```

```python
print("bow Accuracy =>",accuracy_score(y_test, y_pred_nbow))
```

```python
print("tfidf Accuracy =>", accuracy_score(y_test, y_pred_tfidf))
```

```python
print("Bow Confusion matrix :\n", confusion_matrix(y_test, y_pred_nbow))
```

```python
print("tfidf Confusion matrix :\n", confusion_matrix(y_test, y_pred_tfidf))
```

```python
print("bow Classification report: \n", classification_report(y_test, y_pred_nbow))
```

```python
print("tfidf Classification report: \n", classification_report(y_test, y_pred_tfidf))
```
