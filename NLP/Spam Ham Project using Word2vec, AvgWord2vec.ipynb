```python
import pandas as pd
messages = pd.read_csv(r'D:\Machine learning\ML Programs\NLP\SpamClassifier\smsspamcollection\SMSSpamCollection.csv', sep='\t',names=['label','message'])
messages
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
import re
corpus = []
for i in range(0, len(messages)):
    review = re.sub('[^a-zA-Z]', ' ', messages['message'][i])
    review = review.lower()
    review = review.split()
    
    review = [lemmatizer.lemmatize(word) for word in review]
    review = ' '.join(review)
    corpus.append(review)
```

```python
[[i,j,k] for i,j,k in zip(list(map(len,corpus)),corpus, messages['message']) if i<1]
```

```python
corpus
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
words=[]
for sent in corpus:
    sent_token=sent_tokenize(sent)
    for sent in sent_token:
        words.append(simple_preprocess(sent))

words[0]
```

```python
import gensim
```

```python
##lets train word2vec from scrach
model = gensim.models.Word2Vec(words)
```

```python
import numpy as np
def avg_word2vec(doc):
    # remove out-of-vocabulary words
    #sent = [word for word in doc if word in model.wv.index_to_key]
    #print(sent)
    
    return np.mean([model.wv[word] for word in doc if word in model.wv.index_to_key],axis=0)
                #or [np.zeros(len(model.wv.index_to_key))], axis=0)
```

```python
from tqdm import tqdm
```

```python
#apply for the entire sentences
X=[]
for i in tqdm(range(len(words))):
    X.append(avg_word2vec(words[i]))
```

```python
fixed_length = 100
X_fixed = []

for i, x in enumerate(X):
    x = np.array(x).flatten()
    if len(x) < fixed_length:
        # Pad with zeros at the end
        x = np.pad(x, (0, fixed_length - len(x)), mode='constant')
    elif len(x) > fixed_length:
        # Trim extra values
        x = x[:fixed_length]
    X_fixed.append(x)
```

```python
rows = [x.reshape(1, -1) for x in X_fixed]
df = pd.DataFrame(np.vstack(rows))
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestClassifier
classifier = RandomForestClassifier()
```

```python
y_pred = classifier.predict(X_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score, classification_report
print("accuracy =>", accuracy_score(y_test, y_pred) * 100 , "%")
print('classification report :\n ', classification_report(y_test, y_pred))
```
